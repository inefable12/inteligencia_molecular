# 1. Introducción

## ¿Qué es la Inteligencia Artificial (IA)?

La **Inteligencia Artificial** es un campo de la informática que se enfoca en el desarrollo de sistemas y algoritmos capaces de realizar tareas que típicamente requieren de la inteligencia humana. Estas tareas incluyen:

*   **Aprendizaje:** Adquirir información y reglas para usarla.
*   **Razonamiento:** Usar reglas para llegar a conclusiones aproximadas o definitivas.
*   **Autocorrección:** Aprender de los errores y mejorar el rendimiento de manera iterativa.
*   **Percepción:** Interpretar datos sensoriales (imágenes, voz, texto).

En el contexto de la **Inteligencia Molecular** y el *Machine Learning*, la IA nos permite analizar grandes volúmenes de datos químicos y biológicos, identificar patrones en estructuras moleculares, predecir propiedades fisicoquímicas y diseñar nuevos fármacos o materiales de manera eficiente.

```python
#@title Relación Jerárquica de la Inteligencia Artificial

import matplotlib.pyplot as plt

# Configurar la figura y el eje
fig, ax = plt.subplots(figsize=(8, 8))

# Dibujar círculos concéntricos para representar la jerarquía de la IA
# Círculo externo: Inteligencia Artificial
circle_ai = plt.Circle((0.5, 0.5), 0.45, color='#1f77b4', alpha=0.2, label='Inteligencia Artificial (IA)')
ax.add_patch(circle_ai)
ax.text(0.5, 0.88, 'Inteligencia Artificial (IA)', fontsize=14, fontweight='bold', ha='center', va='center', color='#0f3c5a')
ax.text(0.5, 0.82, 'Sistemas que imitan la inteligencia humana\n(Sistemas expertos, lógica difusa, etc.)', fontsize=10, ha='center', va='center', color='#0f3c5a', style='italic')

# Círculo medio: Machine Learning
circle_ml = plt.Circle((0.5, 0.42), 0.32, color='#2ca02c', alpha=0.3, label='Machine Learning (ML)')
ax.add_patch(circle_ml)
ax.text(0.5, 0.64, 'Machine Learning (ML)', fontsize=13, fontweight='bold', ha='center', va='center', color='#165016')
ax.text(0.5, 0.58, 'Algoritmos que aprenden de los datos\n(Regresiones, Árboles, Random Forest)', fontsize=9.5, ha='center', va='center', color='#165016', style='italic')

# Círculo interno: Deep Learning
circle_dl = plt.Circle((0.5, 0.34), 0.18, color='#d62728', alpha=0.4, label='Deep Learning (DL)')
ax.add_patch(circle_dl)
ax.text(0.5, 0.42, 'Deep Learning (DL)', fontsize=12, fontweight='bold', ha='center', va='center', color='#6b1314')
ax.text(0.5, 0.32, 'Redes Neuronales Artificiales\n(CNN, RNN, Transformers)', fontsize=9, ha='center', va='center', color='#6b1314', style='italic')

# Ajustes de visualización
ax.set_xlim(0, 1)
ax.set_ylim(0, 1)
ax.axis('off')
plt.tight_layout()
plt.show()
```

### El Nuevo Paradigma del Machine Learning

**François Chollet**, un destacado ingeniero e investigador de inteligencia artificial en Google y creador de la célebre biblioteca de aprendizaje profundo **Keras**, describe en su libro *"Deep Learning with Python"* un cambio fundamental en la forma de resolver problemas mediante computadoras.

A diferencia del paradigma de la **programación tradicional**, donde los humanos diseñan manualmente las reglas (algoritmos) que procesan los datos para obtener respuestas, el **Machine Learning** invierte esta relación. En este nuevo enfoque, introducimos tanto los datos como las respuestas esperadas (etiquetas) en la máquina, permitiendo que sea el propio sistema el que aprenda y descubra las **reglas** subyacentes. Este cambio paradigmático es lo que permite a las computadoras automatizar tareas complejas que son sumamente difíciles de codificar con reglas explícitas, como el reconocimiento de patrones moleculares o la predicción de propiedades químicas.

El **Aprendizaje Automático en Química** se divide fundamentalmente en tres enfoques de acuerdo a la disponibilidad y uso de los datos: el **Aprendizaje Supervisado**, que utiliza datos etiquetados para la predicción precisa de propiedades moleculares (como el logP o la toxicidad); el **Aprendizaje No Supervisado**, que agrupa y reduce la dimensionalidad de quimiotecas sin necesidad de etiquetas previas para explorar el espacio químico; y el **Aprendizaje por Refuerzo**, que optimiza de forma iterativa y mediante recompensas la generación *de novo* de fármacos viables y la planificación sintética automatizada.

```python
#@title Tipos de Aprendizaje Automático

import matplotlib.pyplot as plt

# Configurar la figura y el eje
fig, ax = plt.subplots(figsize=(10, 8))

# Opciones de estilo
box_style = dict(boxstyle='round,pad=0.5', facecolor='#f5f6f8', edgecolor='#333333', lw=1.5)
main_title_style = dict(boxstyle='round,pad=0.6', facecolor='#4A90E2', edgecolor='#2C5282', lw=2)

# 1. Título Principal
ax.text(0.5, 0.9, "Machine Learning",
        fontsize=16, fontweight='bold', color='white', ha='center', va='center', bbox=main_title_style)

# 2. Ramas Principales
# Supervisado
ax.text(0.18, 0.65, "Aprendizaje Supervisado",
        fontsize=13, fontweight='bold', color='#1A365D', ha='center', va='center', bbox=dict(boxstyle='round,pad=0.4', facecolor='#EBF8FF', edgecolor='#3182CE', lw=1.5))
# No Supervisado
ax.text(0.5, 0.65, "Aprendizaje No Supervisado",
        fontsize=13, fontweight='bold', color='#1C3D27', ha='center', va='center', bbox=dict(boxstyle='round,pad=0.4', facecolor='#F0FFF4', edgecolor='#38A169', lw=1.5))
# Refuerzo
ax.text(0.82, 0.65, "Aprendizaje por Refuerzo",
        fontsize=13, fontweight='bold', color='#5F370E', ha='center', va='center', bbox=dict(boxstyle='round,pad=0.4', facecolor='#FFFFF0', edgecolor='#D69E2E', lw=1.5))

# 3. Aplicaciones Químicas por Categoría
# Supervisado: Ejemplos
supervised_text = (
    "• Predicción de propiedades (QSAR):\n"
    "  Modelos para estimar solubilidad, logP,\n"
    "  afinidad de unión a receptores y toxicidad.\n\n"
    "• Rendimiento de reacciones:\n"
    "  Algoritmos que predicen la eficiencia\n"
    "  de una ruta de síntesis específica.\n\n"
    "• Clasificación de espectros:\n"
    "  Identificación automática de grupos\n"
    "  funcionales en espectros IR o RMN."
)
ax.text(0.18, 0.32, supervised_text,
        fontsize=9.5, color='#2D3748', ha='center', va='center', bbox=dict(boxstyle='round,pad=0.5', facecolor='#FFF', edgecolor='#E2E8F0', lw=1))

# No Supervisado: Ejemplos
unsupervised_text = (
    "• Reducción de dimensionalidad:\n"
    "  Análisis de Componentes Principales (PCA)\n"
    "  o t-SNE para explorar el espacio químico.\n\n"
    "• Clustering de quimiotecas:\n"
    "  Agrupamiento de compuestos por similitud\n"
    "  estructural para cribado virtual diverso.\n\n"
    "• Mapeo de espacio molecular:\n"
    "  Proyección visual de bases de datos\n"
    "  gigantes de ligandos sin etiquetas."
)
ax.text(0.5, 0.32, unsupervised_text,
        fontsize=9.5, color='#2D3748', ha='center', va='center', bbox=dict(boxstyle='round,pad=0.5', facecolor='#FFF', edgecolor='#E2E8F0', lw=1))

# Refuerzo: Ejemplos
reinforcement_text = (
    "• Generación de novo de ligandos:\n"
    "  Optimización de estructuras químicas\n"
    "  premiando la síntesis y penalizando\n"
    "  propiedades no deseadas.\n\n"
    "• Diseño de fármacos optimizados:\n"
    "  Agentes que exploran SMILES buscando\n"
    "  maximizar afinidad y perfil ADMET.\n\n"
    "• Planificación de síntesis (Robótica):\n"
    "  Guiado automatizado de brazos robóticos\n"
    "  de laboratorio mediante prueba y error."
)
ax.text(0.82, 0.32, reinforcement_text,
        fontsize=9.5, color='#2D3748', ha='center', va='center', bbox=dict(boxstyle='round,pad=0.5', facecolor='#FFF', edgecolor='#E2E8F0', lw=1))

# 4. Líneas de Conexión (Flechas)
arrowprops = dict(arrowstyle="->", color="#A0AEC0", lw=1.5)

# De Título a Ramas
ax.annotate('', xy=(0.18, 0.69), xytext=(0.5, 0.85), arrowprops=arrowprops)
ax.annotate('', xy=(0.5, 0.69), xytext=(0.5, 0.85), arrowprops=arrowprops)
ax.annotate('', xy=(0.82, 0.69), xytext=(0.5, 0.85), arrowprops=arrowprops)

# De Ramas a Detalles
ax.annotate('', xy=(0.18, 0.52), xytext=(0.18, 0.61), arrowprops=arrowprops)
ax.annotate('', xy=(0.5, 0.52), xytext=(0.5, 0.61), arrowprops=arrowprops)
ax.annotate('', xy=(0.82, 0.52), xytext=(0.82, 0.61), arrowprops=arrowprops)

# Ajustar límites del gráfico y ocultar ejes
ax.set_xlim(0, 1)
ax.set_ylim(0, 1)
ax.axis('off')

plt.tight_layout()
plt.show()
```

Esta imagen ilustra de forma esquemática las ramas del **Machine Learning** de una manera muy visual:

*   **Aprendizaje Supervisado (Supervised Learning):** Consiste en aprender un mapeo a partir de pares entrada-salida etiquetados (regresión y clasificación).
*   **Aprendizaje No Supervisado (Unsupervised Learning):** Encuentra patrones e información oculta en datos que no contienen etiquetas predefinidas (agrupamiento o *clustering*).
*   **Aprendizaje por Refuerzo (Reinforcement Learning):** Se basa en un agente que toma acciones dentro de un entorno para maximizar una recompensa acumulada mediante el método de prueba y error.
